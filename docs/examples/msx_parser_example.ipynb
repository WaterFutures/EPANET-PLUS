```python
from sympy import symbols

from epanet_plus import msx_parser
```

## Parsing constants

The `parse_constants()` function extracts a constant's name and numerical
value from an MSX coefficient definition.

```python
name, value = msx_parser.parse_constants("COEFFICIENT K1 0.5")
print(f"Constant: {name} = {value}")
```

## Parsing terms

The `parse_terms()` function extracts a term definition and substitutes
any constants found in its expression.

```python
name, expression = msx_parser.parse_terms("T1 K1 * 2", {"K1": 0.5})
print(f"Term: {name} = {expression}")
```

## Parsing a rate equation

The `parse_equation()` function converts an MSX rate equation into an executable Python function.


```python
sym_map = {"A": symbols("A")}
species, rate = msx_parser.parse_equation(
    "RATE A -K1*A",
    {"K1": 2.0},
    {},
    sym_map,
    {},
)

result = rate([3.0], 1.0, 1.0, 1.0)
print(f"Rate equation for {species}: {result}")
```

## Parsing a complete MSX file

The `parse_msx()` function reads an MSX file and collects its species, constants, terms, rate equations, equilibrium equations, and initial conditions.

The following example uses `net2-cl2.msx`.

```python
msx_data = msx_parser.parse_msx("net2-cl2.msx")

print(f"Species: {msx_data['species']}")
print(f"Constants: {msx_data['constants']}")
print(f"Terms: {msx_data['terms']}")
print(f"Rate equations: {list(msx_data['rates'].keys())}")
print(f"Equilibrium equations: {list(msx_data['equilibria'].keys())}")
print(f"Initial conditions: {msx_data['initial_conditions']}")
```
